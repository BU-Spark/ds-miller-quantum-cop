```python
import pandas as pd
```

```python
special2020 = pd.read_csv("Special_Events_Overtime_2012_-_2022.xlsx - 2020.csv")
special2021 = pd.read_csv("Special_Events_Overtime_2012_-_2022.xlsx - 2021.csv")
special2022 = pd.read_csv("Special_Events_Overtime_2012_-_2022.xlsx - 2022.csv")
```

```python
special_combined = pd.concat([special2020, special2021, special2022])
```

```python
special_combined.head()
```

```python
special_combined.shape
```

```python
special_combined.columns
```

```python
special_combined.isna().sum()
```

```python
ok = special_combined[special_combined['ID'].notnull()]
ok = ok[ok.columns[~ok.isnull().all()]]
ok
```

```python
broken = special_combined[special_combined['BRUINS'].notnull()]
```

```python
broken = broken[broken.columns[~broken.isnull().all()]]
broken
```

```python
broken.columns = broken.iloc[-1]
broken = broken[:-1]
broken.reset_index(drop=True)
broken
```

```python
special_fixed = pd.concat([ok, broken])
```

```python
special_fixed.drop(['ASSIGNED', 'CHARGED',])
```

```python
special_fixed.drop(['ASSIGNED', 'CHARGED', 'OTCODE', 'DESCRIPTIONS', 'TYPE'], axis=1, inplace=True)
```

```python
special_fixed.head()
```

```python
special_fixed.to_csv("Special_Events_2020_2021_2022", index=False)
```
