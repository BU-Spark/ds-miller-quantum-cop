```python
import pandas as pd
```

```python
court2020 = pd.read_csv("Copy of Court_Overtime_2020_to_present_-_M_DeMaio.xlsx - Court 2020.csv")
court2021 = pd.read_csv("Copy of Court_Overtime_2020_to_present_-_M_DeMaio.xlsx - Court 2021.csv")
court2022 = pd.read_csv("Copy of Court_Overtime_2020_to_present_-_M_DeMaio.xlsx - Court 2022.csv")
```
