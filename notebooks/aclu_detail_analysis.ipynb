```python
# Compare MassDot overtime with detail overtime

import pandas as pd
massdot = pd.read_csv('../cleaned_data/ACLU_merged_cleaned.csv')
massdot
```

```python
# Import the Detail_merged_cleaned.csv file
detail = pd.read_csv('../cleaned_data/Detail_merged_cleaned.csv')
detail
```

```python
# Convert the 'Emp_No' column in the detail dataframe to a string
detail['Emp_No'] = detail['Emp_No'].astype(str)
```

```python
# Convert the "STARTTIME" column in records to military time like 00:00
detail['Start_Time'] = detail['Start_Time'].astype(str)
detail['Military Start Time'] = detail['Start_Time'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

# Convert the "ENDTIME" column in records to military time like 00:00
detail['End_Time'] = detail['End_Time'].astype(str)
detail['Military End Time'] = detail['End_Time'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")
```

```python
massdot['Event/Written Date'] = pd.to_datetime(massdot['Event/Written Date'])
detail['Start_date'] = pd.to_datetime(detail['Start_date'])

print(detail['Start_date'][0])
print(massdot['Event/Written Date'][0])
print(detail['Military Start Time'][0])
print(massdot['Military Time'][0])
print(detail['Military End Time'][0])

detail['Start_date'][0] == massdot['Event/Written Date'][0]
```

```python
import pandas as pd

# Step 1: Perform a merge to align records based on date and Officer Id
merged_df = pd.merge(
    massdot,
    detail,
    left_on=['Event/Written Date', 'Officer Id'],
    right_on=['Start_date', 'Emp_No'],
    suffixes=('_massdot', '_detail')
)

# Step 2: Filter rows where Military Time is within the specified range
quantum = merged_df[
    (merged_df['Military Start Time'] <= merged_df['Military Time']) &
    (merged_df['Military Time'] <= merged_df['Military End Time'])
][[
    'Officer Id', 'Start_date', 'Event/Written Date', 
    'Military Start Time', 'Military Time', 'Military End Time', 'Issuing Agency', 'District_Worked'
]]

# Reset the index if needed
quantum = quantum.reset_index(drop=True)
```

```python
quantum
```

```python
# Remove any duplicate rows
quantum = quantum.drop_duplicates()

# Reset the index if needed
quantum = quantum.reset_index(drop=True)

quantum
```

```python
# print all unique values in the 'Issuing Agency' column
quantum['Issuing Agency'].unique()
```

```python
quantum['Issuing Agency'] = quantum['Issuing Agency'].str.extract('(\d+)')
quantum['Issuing Agency'] = pd.to_numeric(quantum['Issuing Agency'], errors='coerce').astype('Int64')
quantum
```

```python
# Percent of issuing agency rows and district worked rows that are the same 
same = quantum[quantum['Issuing Agency'] == quantum['District_Worked']]
percent_same = len(same) / len(quantum) * 100
percent_same
```

```python
# Filter out rows where 'Issuing Agency' or 'District_Worked' is None or missing
filtered_quantum = quantum.dropna(subset=['Issuing Agency', 'District_Worked'])

# Percent of issuing agency rows and district worked rows that are the different
diff = filtered_quantum[filtered_quantum['Issuing Agency'] != filtered_quantum['District_Worked']]
percent_diff = len(diff) / len(quantum) * 100
print(percent_diff)
diff
```
