```python
# Compare MassDot overtime with events overtime 

import pandas as pd
massdot = pd.read_csv('../cleaned_data/ACLU_merged_cleaned.csv')
massdot
```

```python
events = pd.read_csv('../cleaned_data/Events_merged_cleaned.csv')
events
```

```python
# Change "ID" from a float64 to an str 8645.0 to 8645 using strip
events['ID'] = events['ID'].astype(str)
events['ID'] = events['ID'].apply(lambda x: x.strip('.0'))
```

```python
# Change "STARTTIME" and "ENDTIME" from a float64 to a str and remove '.0'
events['STARTTIME'] = events['STARTTIME'].astype(str).str.replace('.0', '', regex=False)
events['ENDTIME'] = events['ENDTIME'].astype(str).str.replace('.0', '', regex=False)
```

```python
# Convert the "STARTTIME" column in records to military time like 00:00
events['STARTTIME'] = events['STARTTIME'].astype(str)
events['Military Start Time'] = events['STARTTIME'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

# Convert the "ENDTIME" column in records to military time like 00:00
events['ENDTIME'] = events['ENDTIME'].astype(str)
events['Military End Time'] = events['ENDTIME'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

events
```

```python
massdot['Event/Written Date'] = pd.to_datetime(massdot['Event/Written Date'])
events['OTDATE'] = pd.to_datetime(events['OTDATE'])

print(events['OTDATE'][0])
print(massdot['Event/Written Date'][0])
print(events['Military Start Time'][0])
print(massdot['Military Time'][0])
print(events['Military End Time'][0])

events['OTDATE'][0] == massdot['Event/Written Date'][0]
```

```python
import pandas as pd

# Step 1: Perform a merge to align records based on date and Officer Id
merged_df = pd.merge(
    massdot,
    events,
    left_on=['Event/Written Date', 'Officer Id'],
    right_on=['OTDATE', 'ID'],
    suffixes=('_massdot', '_events')
)

# Step 2: Filter rows where Military Time is within the specified range
quantum = merged_df[
    (merged_df['Military Start Time'] <= merged_df['Military Time']) &
    (merged_df['Military Time'] <= merged_df['Military End Time'])
][[
    'Officer Id', 'OTDATE', 'Event/Written Date', 
    'Military Start Time', 'Military Time', 'Military End Time'
]]

# Reset the index if needed
quantum = quantum.reset_index(drop=True)
```
