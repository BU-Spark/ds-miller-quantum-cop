1. **Agency Code**  
   - **Description**: A unique code representing the issuing law enforcement agency.
   - **Format/Types**: Short alphanumeric code (e.g., "PD_BPH", "CP_BPX"). Each code typically contains initials or abbreviations tied to specific departments or police districts.

2. **Issuing Agency**  
   - **Description**: The name of the law enforcement body responsible for issuing the citation or warning.
   - **Format/Types**: Full or semi-abbreviated text names (e.g., "Boston Police District C-11", "Boston Police Special OPS"). These generally describe a particular police division or district.

3. **Officer Id**  
   - **Description**: A unique identifier or badge number assigned to the officer who issued the citation or warning.
   - **Format/Types**: Numeric or short alphanumeric values (e.g., "131215", "11735"). This is typically a personal or badge ID of the officer.

4. **Citation #**  
   - **Description**: A unique reference number associated with the citation or warning issued.
   - **Format/Types**: Alphanumeric string (e.g., "608180AA", "R8170781"). Each citation has a distinct code to identify it within the law enforcement system.

5. **Type**  
   - **Description**: The type of citation issued, such as a warning or civil violation.
   - **Format/Types**: Short text strings (e.g., "WARN", "CIVIL"). This indicates whether it was a formal citation (e.g., "CIVIL") or an informal warning (e.g., "WARN").

6. **Event/Written Date**  
   - **Description**: The date on which the citation or warning was issued.
   - **Format/Types**: Date formatted as MM/DD/YY (e.g., "1/1/20"). Represents the day the event or citation took place.

7. **Time-HH**  
   - **Description**: The hour part of the time when the citation was issued.
   - **Format/Types**: Two-digit number (e.g., "7", "12"). Uses a 12-hour format, showing the hour of day.

8. **Time-MM**  
   - **Description**: The minute part of the time when the citation was issued.
   - **Format/Types**: Two-digit number (e.g., "10", "45"). This represents the minute within the hour when the citation was issued.

9. **AM-PM**  
   - **Description**: The time designation indicating whether the citation was issued in the morning (AM) or afternoon/evening (PM).
   - **Format/Types**: "AM" or "PM". Specifies the half of the day in the 12-hour time format.

10. **Location Name**  
    - **Description**: The location where the citation or warning was issued.
    - **Format/Types**: Text strings (e.g., "Boston", "Jamaica Plain"). Typically the name of the city or neighborhood where the event took place.

11. **Offense**  
    - **Description**: A code representing the specific offense or violation.
    - **Format/Types**: Numeric or alphanumeric code (e.g., "8980", "894A0"). Often corresponds to a legal code or regulation that has been violated.

12. **Offense Description**  
    - **Description**: A textual description of the offense committed.
    - **Format/Types**: Descriptive text (e.g., "YIELD AT INTERSECTION, FAIL * c89 ¬ß8", "MARKED LANES VIOLATION * c89 ¬ß4A"). This provides the legal violation and possibly the statute number.

13. **Race**  
    - **Description**: The race of the individual who received the citation or warning.
    - **Format/Types**: Short text strings representing racial categories (e.g., "ASIAN", "MIDEST", "BLACK"). 

14. **Gender**  
    - **Description**: The gender of the individual who received the citation or warning.
    - **Format/Types**: Short text strings (e.g., "MALE", "FEMALE"). Specifies the gender of the person issued the citation.

```python
# Save 2020.csv as a dataframe and print the first 5 rows
import pandas as pd
df20 = pd.read_csv('../data/ACLU_2020.csv')
df21 = pd.read_csv('../data/ACLU_2021.csv')
df22 = pd.read_csv('../data/ACLU_2022.csv')

# Print the shape of the dataframes
print(df20.shape)
print(df21.shape)
print(df22.shape)
```

```python
# Print all the columns of the dataframes
print(df20.columns)
print(df21.columns)
print(df22.columns)
```

```python
# Step 1: Rename columns for df21 and df22 to match df20
df21 = df21.rename(columns={
    'fstrAgencyCode': 'Agency Code',
    'fstrIssuingAgency': 'Issuing Agency',
    'fstrOfficerId': 'Officer Id',
    'fstrcitationnumber': 'Citation #',
    'fstrCitationType': 'Type',
    'fdtmEventDate': 'Event/Written Date',
    'fstrTime': 'Time-HH',
    'fstrTimeMinutes': 'Time-MM',
    'fstrAMPM': 'AM-PM',
    'fstrlocation': 'Location ID',
    'fstrLocationName': 'Location Name',
    'fstrChapterSecSub': 'Offense',
    'fstrDispositionDescription': 'Offense Description',
    'fstrRace': 'Race',
    'fstrSex': 'Gender'
})

df22 = df22.rename(columns={
    'Issuing Agency Code': 'Agency Code',
    'OfficerID': 'Officer Id',
    'Citation #': 'Citation #',
    'Citation Type': 'Type',
    'Event Date': 'Event/Written Date',
    'Time-HH': 'Time-HH',
    'Time-MM': 'Time-MM',
    'AM-PM': 'AM-PM',
    'Location Name': 'Location Name', # Location ID is not available in this dataset, can be added ~
    'Offense': 'Offense',
    'Offense Description': 'Offense Description',
    'Race': 'Race',
    'Gender': 'Gender'
})

# Step 2: Standardize the 'Event/Written Date' column format
df20['Event/Written Date'] = pd.to_datetime(df20['Event/Written Date'], format="%Y-%m-%dT%H:%M:%SZ").dt.strftime('%Y-%m-%d')
df21['Event/Written Date'] = pd.to_datetime(df21['Event/Written Date'], format="%m/%d/%y %H:%M", errors='coerce').dt.strftime('%Y-%m-%d')
df22['Event/Written Date'] = pd.to_datetime(df22['Event/Written Date'], format="%m/%d/%Y %I:%M:%S %p", errors='coerce').dt.strftime('%Y-%m-%d')

# Step 3: Select the relevant columns
common_columns = ['Agency Code', 'Issuing Agency', 'Officer Id', 'Citation #', 'Type', 
                  'Event/Written Date', 'Time-HH', 'Time-MM', 'AM-PM', 
                  'Location Name', 'Offense', 'Offense Description', 'Race', 'Gender'] # Did not include 'year' as it is already included in Event/Written Date

df20 = df20[common_columns]  # df20 already has the right columns
df21 = df21[common_columns]  # select only the common columns from df21
df22 = df22[common_columns]  # select only the common columns from df22

# Step 4: Concatenate the dataframes
combined_df = pd.concat([df20, df21, df22], ignore_index=True)
```

```python
import pandas as pd

# Assuming combined_df is already created and contains the necessary data

# Step 1: Define the lists for agency codes and issuing agencies
agency_codes = [
    'PD_BPL', 'PD_BPH', 'PD_BPC', 'PD_BPD', 'PD_BPB', 
    'PD_BPE', 'CP_BPX', 'PD_BPK', 'PD_BPA', 'PD_BPF', 
    'PD_BPJ', 'PD_BPG', 'PD_BPX', 'PD_664', 'PD_658', 
    'PD_665', 'PD_654', 'PD_661', 'PD_663', 'PD_666', 
    'PD_667', 'PD_655', 'PD_669', 'PD_662'
]

issuing_agencies = [
    'Boston Police District E-18', 'Boston Police District C-11',
    'Boston Police District B-3', 'Boston Police District D-4',
    'Boston Police District B-2', 'Boston Police District E-5', 
    'Boston Police District D-14', 'Boston Police District A-1',
    'Boston Police District C-6', 'Boston Police District E-13',
    'Boston Police District A-7', 'Boston Police Special OPS',
    'Boston Police Area B', 'Boston Police Area D', 
    'Boston Police Area A', 'Boston Police Area C',
    'Boston Police Area K', 'Boston Police Area H',
    'Boston Police Area L', 'Boston Police Area F',
    'Boston Police Area G', 'Boston Police Area J', 
    'Boston Police Area E'
]

    # ' ',  # Note: You might want to handle the empty string separately

# Step 2: Filter the DataFrame
filtered_combined_df = combined_df[
    (combined_df['Agency Code'].isin(agency_codes)) | 
    (combined_df['Issuing Agency'].isin(issuing_agencies))
]

# Remove all rows in DataFrame that have a NaN value in the "Time-HH" column or NaN value in the "Officer Id" column
filtered_combined_df = filtered_combined_df.dropna(subset=['Time-HH', 'Officer Id'])

# Drop rows that have "UNK" in the "AM-PM" column   
filtered_combined_df = filtered_combined_df[filtered_combined_df['AM-PM'] != 'UNK']

# Add military time based on the "Time-HH", "Time-MM", and AM/PM columns in massdot like 00:00, or 23:59
filtered_combined_df['Time-HH'] = filtered_combined_df['Time-HH'].astype(int)
filtered_combined_df['Time-MM'] = filtered_combined_df['Time-MM'].astype(int)
filtered_combined_df['AM-PM'] = filtered_combined_df['AM-PM'].astype(str)

# Adjust hours for PM times
filtered_combined_df['Military Time'] = filtered_combined_df.apply(
    lambda x: f"{(x['Time-HH'] % 12 + (12 if x['AM-PM'] == 'PM' else 0)):02d}:{x['Time-MM']:02d}", axis=1
)

# Remove duplicate rows
filtered_combined_df = filtered_combined_df.drop_duplicates()

# Step 5: Save the filtered combined dataframe to a CSV file
filtered_combined_df.to_csv('../cleaned_data/ACLU_merged_cleaned.csv', index=False)
```
