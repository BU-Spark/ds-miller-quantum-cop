```python
# Import the Court_merged_cleaned.csv file

import pandas as pd
court = pd.read_csv('../cleaned_data/Court_merged_cleaned.csv')
court
```

```python
# Import the Detail_merged_cleaned.csv file
detail = pd.read_csv('../cleaned_data/Detail_merged_cleaned.csv')
detail
```

```python
court['OTDATE'] = pd.to_datetime(court['OTDATE'])

# Change "IDNO6" from a float64 to an object 8645.0 to 8645 using strip
court['IDNO6'] = court['IDNO6'].astype(str)
court['IDNO6'] = court['IDNO6'].apply(lambda x: x.strip('.0'))

# Convert the "STARTTIME" column in records to military time like 00:00
court['STARTTIME'] = court['STARTTIME'].astype(str)
court['Court Military Start Time'] = court['STARTTIME'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

# Convert the "ENDTIME" column in records to military time like 00:00
court['ENDTIME'] = court['ENDTIME'].astype(str)
court['Court Military End Time'] = court['ENDTIME'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")
```

```python
# Create an actual court paid end time column because the paid end time is always 4 hours after the start time
# Take the max of the (Court Military Start Time + 4 hours) and the Court Military End Time

court['Court Military Start Time'] = pd.to_datetime(court['Court Military Start Time'])
court['Court Military End Time'] = pd.to_datetime(court['Court Military End Time'])

# Max of Court Military Start Time + pd.DateOffset(hours=4) and Court Military End Time
court['Court Military Start Time + 4 hours'] = court['Court Military Start Time'] + pd.DateOffset(hours=4)
court['Court Paid End Time'] = court[['Court Military Start Time + 4 hours', 'Court Military End Time']].max(axis=1)
```

```python
detail['Start_date'] = pd.to_datetime(detail['Start_date'])

# Convert the 'Emp_No' column in the detail dataframe to a string
detail['Emp_No'] = detail['Emp_No'].astype(str)

# Convert the "STARTTIME" column in records to military time like 00:00
detail['Start_Time'] = detail['Start_Time'].astype(str)
detail['Detail Military Start Time'] = detail['Start_Time'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

# Convert the "ENDTIME" column in records to military time like 00:00
detail['End_Time'] = detail['End_Time'].astype(str)
detail['Detail Military End Time'] = detail['End_Time'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")
```

```python
import pandas as pd

# Step 1: Perform a merge to align records based on date and Officer Id
merged_df = pd.merge(
    detail,
    court,
    left_on=['Start_date', 'Emp_No'],
    right_on=['OTDATE', 'IDNO6'],
    suffixes=('_massdot', '_court')
)

# Step 2: Filter rows where Military Time is within the specified range
quantum = merged_df[
    (merged_df['Court Military Start Time'] < merged_df['Detail Military Start Time']) &
    (merged_df['Detail Military Start Time'] < merged_df['Court Paid End Time'])
][[
    'Emp_No', 'OTDATE', 'Start_date', 
    'Court Military Start Time', 'Detail Military Start Time', 'Court Military End Time', 'Court Paid End Time', 'Detail Military End Time'
]]

# Reset the index if needed
quantum = quantum.reset_index(drop=True)
```

```python
events = pd.read_csv('../cleaned_data/Events_merged_cleaned.csv')
events
```

```python
events['OTDATE'] = pd.to_datetime(events['OTDATE'])

# Change "ID" from a float64 to an str 8645.0 to 8645 using strip
events['ID'] = events['ID'].astype(str)
events['ID'] = events['ID'].apply(lambda x: x.strip('.0'))

# Change "STARTTIME" and "ENDTIME" from a float64 to a str and remove '.0'
events['STARTTIME'] = events['STARTTIME'].astype(str).str.replace('.0', '', regex=False)
events['ENDTIME'] = events['ENDTIME'].astype(str).str.replace('.0', '', regex=False)

# Convert the "STARTTIME" column in records to military time like 00:00
events['STARTTIME'] = events['STARTTIME'].astype(str)
events['Events Military Start Time'] = events['STARTTIME'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

# Convert the "ENDTIME" column in records to military time like 00:00
events['ENDTIME'] = events['ENDTIME'].astype(str)
events['Events Military End Time'] = events['ENDTIME'].apply(lambda x: '00:00' if x == '0' else f"{x.zfill(4)[:2]}:{x.zfill(4)[2:]}")

events
```

```python
import pandas as pd

# Step 1: Perform a merge to align records based on date and Officer Id
merged_df = pd.merge(
    events,
    court,
    left_on=['OTDATE', 'ID'],
    right_on=['OTDATE', 'IDNO6'],
    suffixes=('_massdot', '_court')
)

# Step 2: Filter rows where Military Time is within the specified range
quantum = merged_df[
    (merged_df['Court Military Start Time'] < merged_df['Events Military Start Time']) &
    (merged_df['Events Military Start Time'] < merged_df['Court Paid End Time'])
][[
    'ID', 'OTDATE',
    'Court Military Start Time', 'Events Military Start Time', 'Court Military End Time', 'Court Paid End Time', 'Events Military End Time'
]]

# Reset the index if needed
quantum = quantum.reset_index(drop=True)
```
